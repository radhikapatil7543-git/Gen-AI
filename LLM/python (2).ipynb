```python
import google.genai as genai
print(genai.__version__)
```

```python
import pathlib
import textwrap

from IPython.display import display, Markdown
from google.colab import userdata

def to_markdown(text):
  text = text.replace('•', '  *')
  return Markdown(textwrap.indent(text, '> ', predicate=lambda _: True))
```

```python
GOOGLE_API_KEY=userdata.get('GOOGLE_API_KEY')
```

```python
client = genai.Client(api_key = GOOGLE_API_KEY)
print("Client created successfully!")
```

```python
for model in client.models.list():
  print(model.name)
```

```python
response = client.models.generate_content(model = 'gemini-3.5-flash', contents = 'What is the meaning of life?')
print(response.text)
```

```python
response = client.models.generate_content(model = 'gemini-3.5-flash', contents = 'What is the meaning of life?')
display(to_markdown(response.text))
```

```python
response = client.models.generate_content(model = 'gemini-3.5-flash', contents = 'What is the meaning of life?', config = {'response_modalities': ['TEXT']})
print(response.text)
```

```python
import PIL.Image

img = PIL.Image.open('lotus.jpg')
display(img)
```

```python
response = client.models.generate_content(model = 'gemini-3.5-flash', contents = [img])
display(to_markdown(response.text))
```

```python
response = client.models.generate_content(model = 'gemini-3.5-flash', contents = ['Write a short blog post about this flower.', img])
display(to_markdown(response.text))
```
